# 03 Clinician ratings

H4 (edit time vs omission vs hallucination counts) is testable only after real reviews exist under `data/human_eval/`. Scaffold CLI rows are not clinician data. Do not report ICC until N≥3 reviewers complete `docs/clinician_protocol.md`.

```python
import json
from pathlib import Path

files = list(Path('data/human_eval').glob('*.jsonl'))
rows = []
for path in files:
    for line in path.read_text().splitlines():
        if line.strip():
            rows.append(json.loads(line))
print('review rows', len(rows))
if not rows:
    print('No clinician data. Run the protocol; do not invent scores.')
else:
    print('If these rows have notes==scaffold, they are not usable for H4.')
```
