# 01 Error analysis

Seed annotation files are **programmatic** copies of planted tokens. Cohen's κ on those files is not human IAA. Replace `data/annotations/` after dual annotation and adjudication.

```python
from collections import Counter
import json
from pathlib import Path

def load(name):
    path = Path('data/annotations') / name
    rows = []
    for line in path.read_text().splitlines():
        if line.strip():
            rows.append(json.loads(line))
    return rows

a1, a2 = load('A1.jsonl'), load('A2.jsonl')
print('A1', len(a1), 'A2', len(a2))
print('error_type counts A1', Counter(r['error_type'] for r in a1))
print('These counts are seed labels, not human κ.')
```
