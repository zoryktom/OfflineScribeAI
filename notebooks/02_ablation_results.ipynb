# 02 Ablation results

Reads the newest `experiments/runs/*/metrics.json` if present. Stub runs are seeded perturbations, not model benchmarks.

```python
import json
from pathlib import Path

runs = sorted(Path('experiments/runs').glob('*/metrics.json'))
if not runs:
    print('No runs yet. Execute: offlinescribe eval run --config experiments/configs/sample.yaml')
else:
    payload = json.loads(runs[-1].read_text())
    print(runs[-1])
    print(payload.get('headline_note'))
    for name, cell in payload.get('by_condition', {}).items():
        h = cell['hallucination_rate']
        o = cell['omission_rate']
        print(f"{name}: hallu={h['mean']:.3f} [{h['ci95_lo']:.3f},{h['ci95_hi']:.3f}] omit={o['mean']:.3f}")
```
